```python
import sys
import os
sys.path.insert(0, os.path.abspath('../'))

import techind as ti
import pandas as pd

df = pd.read_pickle('../data/test_data01.pkl')
```

```python
df['TEMA'] = ti.trippleExponentialMovingAverage(30, df['adjclose'].values)
df['WMA'] = ti.weightedMovingAverage(30, df['adjclose'].values)
df['EMA'] = ti.exponentialMovingAverage(30, df['adjclose'].values)
df['TMA'] = ti.triangularMovingAverage(30, df['adjclose'].values)

df.tail()
```

```python
df.loc[df.index > pd.to_datetime("2018-01-01").date()].plot(figsize=(16,10))
```
